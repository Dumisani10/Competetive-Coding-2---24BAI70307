```python
from typing import List


class Solution:
    def findDuplicate(self, nums: List[int]) -> int:
        
        slow = nums[0]
        fast = nums[0]

        while True:
            slow = nums[slow]
            fast = nums[nums[fast]]

            if slow == fast:
                break

        slow = nums[0]

        while slow != fast:
            slow = nums[slow]
            fast = nums[fast]

        return slow


nums = list(map(int, input("Enter numbers: ").split()))

solution = Solution()

result = solution.findDuplicate(nums)

print("Duplicate number:", result)
```
