```python

def subsets(nums):
    result = []
    n = len(nums)

    for mask in range(0, 2 ** n):
        subset = []

        for i in range(n):
            if mask & (1 << i):
                subset.append(nums[i])

        result.append(subset)

    return result


nums = [1, 2, 3]
print(subsets(nums))
```
