```python
def AddDigits(num):
    while num >= 10:
        total = 0
        while num > 0:
            total += num % 10
            num //= 10
        num = total
    return num

num = int(input("Enter a number: "))
result = AddDigits(num)
print(f"Adding the digits of {num} until a single digit is obtained: {result}")
```
