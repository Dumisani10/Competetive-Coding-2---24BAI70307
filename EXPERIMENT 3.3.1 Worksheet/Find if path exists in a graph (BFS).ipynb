```python
from collections import deque


def valid_path(n, edges, source, destination):
    adj = [[] for _ in range(n)]

    for u, v in edges:
        adj[u].append(v)
        adj[v].append(u)  

    if source == destination:
        return True

    visited = [False] * n

    queue = deque()

    queue.append(source)
    visited[source] = True

    while queue:
        node = queue.popleft()

        for neighbour in adj[node]:

            if neighbour == destination:
                return True

            if not visited[neighbour]:
                visited[neighbour] = True
                queue.append(neighbour)

    return False


n = 3
edges = [[0, 1], [1, 2], [2, 0]]
source = 0
destination = 2

print("Test Case 1:", valid_path(n, edges, source, destination))

n = 6
edges = [[0, 1], [0, 2], [3, 5], [5, 4], [4, 3]]
source = 0
destination = 5

print("Test Case 2:", valid_path(n, edges, source, destination))

n = 1
edges = []
source = 0
destination = 0

print("Test Case 3:", valid_path(n, edges, source, destination))
```
