```python

class ListNode:
    def __init__(self, data):
        self.data = data
        self.next = None


def find_cycle_start(head):

    slow = head
    fast = head

    while fast is not None and fast.next is not None:

        slow = slow.next
        fast = fast.next.next

        if slow == fast:
            break

    else:
        return None

  
    slow = head

    while slow != fast:

        slow = slow.next
        fast = fast.next

    return slow


head = ListNode(1)
node2 = ListNode(2)
node3 = ListNode(3)
node4 = ListNode(4)
node5 = ListNode(5)

head.next = node2
node2.next = node3
node3.next = node4
node4.next = node5


node5.next = node3


start = find_cycle_start(head)

if start is not None:
    print("Cycle starts at node:", start.data)
else:
    print("No cycle found")
```
