```python

from collections import deque


class StackUsingQueue:
    def __init__(self):
        self.queue = deque()

    
    def push(self, value):
        self.queue.append(value)

        
        for _ in range(len(self.queue) - 1):
            self.queue.append(self.queue.popleft())

        print(value, "pushed into stack")

    
    def pop(self):
        if not self.queue:
            print("Stack is empty")
            return

        value = self.queue.popleft()
        print(value, "popped from stack")

    
    def peek(self):
        if not self.queue:
            print("Stack is empty")
            return

        print("Top element:", self.queue[0])

 
    def display(self):
        if not self.queue:
            print("Stack is empty")
            return

        print("Stack:", list(self.queue))



s = StackUsingQueue()

while True:
    print("\n--- Stack Using Queue ---")
    print("1. Push")
    print("2. Pop")
    print("3. Peek")
    print("4. Display")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        value = input("Enter value: ")
        s.push(value)

    elif choice == "2":
        s.pop()

    elif choice == "3":
        s.peek()

    elif choice == "4":
        s.display()

    elif choice == "5":
        print("Program ended.")
        break

    else:
        print("Invalid choice")
```
