```python

class QueueUsingStacks:
    def __init__(self):
        self.stack1 = []
        self.stack2 = []

    
    def enqueue(self, value):
        self.stack1.append(value)
        print(value, "added to queue")

    
    def dequeue(self):
        if not self.stack1 and not self.stack2:
            print("Queue is empty")
            return

        
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())

        value = self.stack2.pop()
        print(value, "removed from queue")

   
    def peek(self):
        if not self.stack1 and not self.stack2:
            print("Queue is empty")
            return

        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())

        print("Front element:", self.stack2[-1])

   
    def display(self):
        if not self.stack1 and not self.stack2:
            print("Queue is empty")
            return

        queue = self.stack2[::-1] + self.stack1
        print("Queue:", queue)



q = QueueUsingStacks()

while True:
    print("\n--- Queue Using Stacks ---")
    print("1. Enqueue")
    print("2. Dequeue")
    print("3. Peek")
    print("4. Display")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        value = input("Enter value: ")
        q.enqueue(value)

    elif choice == "2":
        q.dequeue()

    elif choice == "3":
        q.peek()

    elif choice == "4":
        q.display()

    elif choice == "5":
        print("Program ended.")
        break

    else:
        print("Invalid choice")

```
