```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def createLinkedList(arr):
    if not arr:
        return None

    head = ListNode(arr[0])
    current = head

    for x in arr[1:]:
        current.next = ListNode(x)
        current = current.next

    return head


def isPalindrome(head):
    arr = []

    while head:
        arr.append(head.val)
        head = head.next

    return arr == arr[::-1]


head = createLinkedList([1, 2, 2, 1])

print(isPalindrome(head))
```
