```python

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:

    def inorderSuccessor(self, root, p):


        successor = None

        while root is not None:


            if p.val >= root.val:
                root = root.right

            else:

                successor = root

                root = root.left

        return successor


root = TreeNode(5)

root.left = TreeNode(3)
root.right = TreeNode(6)

root.left.left = TreeNode(2)
root.left.right = TreeNode(4)


p = root.left       



solution = Solution()

result = solution.inorderSuccessor(root, p)



print("Node p:", p.val)

if result is not None:
    print("Inorder Successor:", result.val)
else:
    print("Inorder Successor: None")
```
