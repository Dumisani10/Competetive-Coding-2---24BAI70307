```python

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:

    def inorderSuccessor(self, root, p):

        order = []

        def inorder(node):
            if node is None:
                return

            inorder(node.left)

            order.append(node)

            inorder(node.right)

        inorder(root)

        for i in range(len(order)):

            if order[i] == p:

                if i + 1 < len(order):
                    return order[i + 1]

                return None

        return None


root = TreeNode(5)

root.left = TreeNode(3)
root.right = TreeNode(7)

root.left.left = TreeNode(2)
root.left.right = TreeNode(4)

root.right.left = TreeNode(6)
root.right.right = TreeNode(8)



p = root.left       


solution = Solution()

result = solution.inorderSuccessor(root, p)

print("Node p:", p.val)

if result is not None:
    print("Inorder Successor:", result.val)
else:
    print("Inorder Successor: None")
```
