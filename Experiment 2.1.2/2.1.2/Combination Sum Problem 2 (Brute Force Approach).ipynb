```python
#Brute force solution to the combination sum problem
def solve(remaining, current, candidates, result):

    if remaining == 0:
        result.add(tuple(sorted(current)))
        return

    if remaining < 0:
        return

   
    for c in candidates:
        current.append(c)

        solve(
            remaining - c,
            current,
            candidates,
            result
        )

        current.pop()


def combination_sum(candidates, target):
    result = set()

    solve(
        target,
        [],
        candidates,
        result
    )

    return [list(combination) for combination in result]



candidates = [2, 3, 6, 7]
target = 7

answer = combination_sum(candidates, target)

print("Combinations:", answer)
```
