```python

def solve(i, height, k):
    if i == 0:
        return 0

    best = float('inf')

    for j in range(max(0, i - k), i):

        jump_cost = abs(height[i] - height[j])

        best = min(best, solve(j, height, k) + jump_cost)

    return best

height = list(map(int, input("Enter heights: ").split()))

k = int(input("Enter maximum jump distance k: "))

result = solve(len(height) - 1, height, k)

print("Minimum cost:", result)
```
