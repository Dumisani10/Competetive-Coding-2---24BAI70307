```python
#Output 1
def ProductOAEself(nums):
    ans = [1] * len(nums)

    for i in range(len(nums)):
        p = 1
        for j in range(len(nums)):
            if i != j:
                p *= nums[j]
        ans[i] = p

    return ans


nums = [1, 2, 3, 4]
print(ProductOAEself(nums))
```
