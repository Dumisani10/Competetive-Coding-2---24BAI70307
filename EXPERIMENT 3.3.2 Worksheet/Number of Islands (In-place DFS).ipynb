```python
def num_islands(grid):

    if not grid:
        return 0

    m = len(grid)
    n = len(grid[0])

    count = 0

    def dfs(r, c):

        if r < 0 or r >= m or c < 0 or c >= n:
            return

        if grid[r][c] != '1':
            return

        grid[r][c] = '0'

        dfs(r - 1, c)  
        dfs(r + 1, c)  
        dfs(r, c - 1)  
        dfs(r, c + 1) 

    for r in range(m):
        for c in range(n):

            if grid[r][c] == '1':
                count += 1
                dfs(r, c)

    return count


m = int(input("Enter number of rows: "))
n = int(input("Enter number of columns: "))

grid = []

print("Enter the grid:")
for i in range(m):
    row = input().split()
    grid.append(row)

result = num_islands(grid)

print("Number of islands:", result)
```
